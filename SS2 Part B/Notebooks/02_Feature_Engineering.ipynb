```python
import numpy as np
import pandas as pd
from pathlib import Path
import matplotlib.pyplot as plt
from sklearn.model_selection import train_test_split
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder, StandardScaler
from sklearn.pipeline import Pipeline
from sklearn.linear_model import LogisticRegression
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score, roc_auc_score, classification_report, confusion_matrix, ConfusionMatrixDisplay, RocCurveDisplay
RANDOM_STATE=42
DATA_DIR=Path("../data"); RESULTS_DIR=Path("../results")
DATA_DIR.mkdir(parents=True,exist_ok=True); RESULTS_DIR.mkdir(parents=True,exist_ok=True)
```

```python
LOCAL_PATH=DATA_DIR/"Telco-Customer-Churn.csv"
PUBLIC_URL="https://raw.githubusercontent.com/IBM/telco-customer-churn-on-icp4d/master/data/Telco-Customer-Churn.csv"
if LOCAL_PATH.exists():
    df=pd.read_csv(LOCAL_PATH)
else:
    df=pd.read_csv(PUBLIC_URL); df.to_csv(LOCAL_PATH,index=False)
print("Shape:",df.shape)
display(df.head())
```

```python
df=df.copy()
df["TotalCharges"]=pd.to_numeric(df["TotalCharges"],errors="coerce")
df=df.dropna(subset=["TotalCharges"]).reset_index(drop=True)
df["Churn"]=df["Churn"].map({"No":0,"Yes":1}).astype(int)
df=df.drop(columns=["customerID"])
print("Cleaned shape:",df.shape)
display(df["Churn"].value_counts())
```

```python
service_cols=["PhoneService","MultipleLines","OnlineSecurity","OnlineBackup","DeviceProtection","TechSupport","StreamingTV","StreamingMovies"]
service_cols=[c for c in service_cols if c in df.columns]
df["service_count"]=df[service_cols].apply(lambda c:(c=="Yes").astype(int)).sum(axis=1)
df["has_internet"]=(df["InternetService"]!="No").astype(int)
df["has_phone"]=(df["PhoneService"]=="Yes").astype(int)
df["monthly_charge_per_tenure"]=np.where(df["tenure"]>0,df["MonthlyCharges"]/df["tenure"],0)
df["tenure_band"]=pd.cut(df["tenure"],bins=[-1,6,12,24,48,np.inf],labels=["0-6","7-12","13-24","25-48","49+"])
display(df[["tenure","MonthlyCharges","service_count","has_internet","has_phone","monthly_charge_per_tenure","tenure_band","Churn"]].head())
```

```python
df.to_csv(DATA_DIR/"engineered_telco_churn.csv",index=False)
print("Saved engineered dataset.")
```
