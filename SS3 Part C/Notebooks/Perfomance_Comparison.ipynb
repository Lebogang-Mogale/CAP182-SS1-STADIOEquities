```python
#Import necessary Liabraries
import pandas as pd
import matplotlib.pyplot as plt
from pathlib import Path

MODEL1_FILE = "Model1Performance_results.csv"
MODEL2_FILE = "Model2Performance_results.csv"

print("Model 1 results file exists:", Path(MODEL1_FILE).exists())
print("Model 2 results file exists:", Path(MODEL2_FILE).exists())
```

## 1. Load the two model results

```python
model1 = pd.read_csv(MODEL1_FILE)
model2 = pd.read_csv(MODEL2_FILE)

model1["Model"] = "Model 1 – Logistic Regression"
model2["Model"] = "Model 2 – Random Forest"

print("Model 1 results:")
display(model1)

print("Model 2 results:")
display(model2)
```

## 2. Create the comparison table

```python
comparison = pd.DataFrame({
    "Metric": model1["Metric"],
    "Model 1 – Logistic Regression": model1["Score"],
    "Model 2 – Random Forest": model2["Score"]
})

comparison["Difference (Model 2 - Model 1)"] = (
    comparison["Model 2 – Random Forest"]
    - comparison["Model 1 – Logistic Regression"]
)

display(
    comparison.style.format({
        "Model 1 – Logistic Regression": "{:.4f}",
        "Model 2 – Random Forest": "{:.4f}",
        "Difference (Model 2 - Model 1)": "{:+.4f}"
    })
)
```

## 3. Percentage comparison

```python
comparison_percentage = comparison.copy()

comparison_percentage["Model 1 – Logistic Regression"] *= 100
comparison_percentage["Model 2 – Random Forest"] *= 100
comparison_percentage["Difference (Model 2 - Model 1)"] *= 100

display(
    comparison_percentage.style.format({
        "Model 1 – Logistic Regression": "{:.2f}%",
        "Model 2 – Random Forest": "{:.2f}%",
        "Difference (Model 2 - Model 1)": "{:+.2f} percentage points"
    })
)
```

## 4. Visual comparison of the five metrics

```python
plot_df = comparison.set_index("Metric")[[
    "Model 1 – Logistic Regression",
    "Model 2 – Random Forest"
]]

ax = plot_df.plot(kind="bar", figsize=(10, 6))
ax.set_title("Model 1 vs Model 2 Performance")
ax.set_xlabel("Metric")
ax.set_ylabel("Score")
ax.set_ylim(0, 1)
ax.legend(title="Model")
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```

## 5. Identify the higher observed score for each metric

```python
rows = []

for _, row in comparison.iterrows():
    m1 = row["Model 1 – Logistic Regression"]
    m2 = row["Model 2 – Random Forest"]

    if m1 > m2:
        higher = "Model 1 – Logistic Regression"
    elif m2 > m1:
        higher = "Model 2 – Random Forest"
    else:
        higher = "Equal"

    rows.append({
        "Metric": row["Metric"],
        "Model 1": m1,
        "Model 2": m2,
        "Higher observed score": higher
    })

metric_summary = pd.DataFrame(rows)

display(
    metric_summary.style.format({
        "Model 1": "{:.4f}",
        "Model 2": "{:.4f}"
    })
)
```

## 7. Save the comparison

```python
comparison.to_csv("Comparison_results.csv", index=False)
print("Saved: Comparison_results.csv")
```
